```python
import numpy as np
import pandas as pd
```

```python
#Load Data
df=pd.read_csv("/content/HR-Employee.csv")
df
```

```python
df.info()
```

```python
df.shape
```

```python
#Drop unneeded columns
columns_to_drop=['EmployeeCount','Over18','StandardHours']
df.drop(columns=columns_to_drop,inplace=True,errors='ignore')
df
```

```python
#convert Attrition to Numeric(0/1)
df['Attrition']=df['Attrition'].apply(lambda x:1 if x=='Yes' else 0)
```

```python
#Create Age Group column
def classify_age(age):
    if age<30:
        return '18-29 (Youth)'
    elif age<49:
        return '30-49 (Mid-Career)'
    else:
        return '50+ (Senior)'
df['Age_Group']=df['Age'].apply(classify_age)
```

```python
#Trim whitespace in text columns(if any)
text_cols=df.select_dtypes(include=['object']).columns
for col in text_cols:
    df[col]=df[col].astype(str).str.strip()
```

```python
#Save cleaned file
df.to_csv("cleaned_hr_analytics.csv",index=False)
print("\n Data Cleaned Successfully! File saved as :cleaned_hr_analytics.csv ")
```
